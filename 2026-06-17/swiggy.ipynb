```python
import pandas as pd

# load the parquet file into a pandas DataFrame
df = pd.read_parquet("swiggy_instamart_2026_06_16.parquet")
print(df.head(5))
print(df.columns.tolist())

```

```python
#convert the DataFrame to a CSV file
df.to_csv("swiggy_instamart_2026_06_16.csv", index=False)
```

```python
#convert the DataFrame to a JSON file
df.to_json("swiggy_instamart_2026_06_16.json", orient="records", lines=True)
```

```python
#convert the DataFrame to an Excel file
df.to_excel("swiggy_instamart_2026_06_16.xlsx", index=False, engine='openpyxl')
```

```python
print("QA Report:")
print("----------------")



print("Columns :")
print("----------------")
print(df.columns.tolist())



print("\nNull Summary :")
print("----------------")
null_summary = pd.DataFrame({
    'Null_Count': df.isnull().sum(),
    'Null_Percentage': (df.isnull().sum() / len(df)) * 100
})
null_summary['Null_Percentage'] = null_summary['Null_Percentage'].round(2)
print(null_summary)



print("\nCompletely Null Columns :")
print("----------------")
completely_null_columns = null_summary[null_summary['Null_Percentage'] == 100].index.tolist()
print(f"Columns that are completely null: {completely_null_columns}")
print(f"Number of completely null columns: {len(completely_null_columns)}")


```

```python
# Check if unique_id and product_url are not empty and unique
print("Validation Check: unique_id and product_url\n")
print("=" * 60)

# Check unique_id
print("\n📌 unique_id:")
unique_id_empty = df['unique_id'].isnull().sum() + (df['unique_id'].astype(str).str.strip() == '').sum()
unique_id_count = len(df['unique_id'].unique())
total_rows = len(df)

print(f"   Empty values: {unique_id_empty}")
print(f"   Total rows: {total_rows}")
print(f"   Unique values: {unique_id_count}")
print(f"   Is NOT empty: {'✓ YES' if unique_id_empty == 0 else '✗ NO'}")
print(f"   Is UNIQUE: {'✓ YES' if unique_id_count == total_rows else f'✗ NO ({total_rows - unique_id_count} duplicates)'}")

# Check product_url
print("\n📌 product_url:")
product_url_empty = df['product_url'].isnull().sum() + (df['product_url'].astype(str).str.strip() == '').sum()
product_url_count = len(df['product_url'].unique())

print(f"   Empty values: {product_url_empty}")
print(f"   Total rows: {total_rows}")
print(f"   Unique values: {product_url_count}")
print(f"   Is NOT empty: {'✓ YES' if product_url_empty == 0 else '✗ NO'}")
print(f"   Is UNIQUE: {'✓ YES' if product_url_count == total_rows else f'✗ NO ({total_rows - product_url_count} duplicates)'}")

print("\n" + "=" * 60)
```
