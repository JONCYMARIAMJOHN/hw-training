```python
# dataframe to parquet

import pandas as pd
import pyarrow as pa

df = pd.DataFrame({
    "unique_id": [1001, 1002, 1003, 1004, 1005],
    "product_name": [
        "Apple iPhone 15",
        "Samsung Galaxy S24",
        "Sony WH-1000XM5",
        "Dell Inspiron 15",
        "Logitech MX Master 3S"
    ],
    "category": [
        "Mobile",
        "Mobile",
        "Headphones",
        "Laptop",
        "Mouse"
    ],
    "selling_price": [799.99, 749.99, 299.99, 649.50, 89.99],
    "regular_price": [899.99, 849.99, 349.99, 699.50, 99.99],
    "in_stock": [True, True, False, True, True]
})

df.to_parquet("sample_data.parquet", index=False)
```

```python
# reading parquet to dataframe

loaded_df = pd.read_parquet("sample_data.parquet")
print(loaded_df)
```

```python
# dataframe to CSV and JSON

loaded_df.to_csv("sample.csv", index=False)

# Save as JSON
loaded_df.to_json("sample.json", orient="records", indent=4)
```

```python
import pyarrow.csv as csv
import pyarrow.parquet as pq

# Read CSV
table = csv.read_csv("sample.csv")

# Write Parquet
pq.write_table(table, "sample_converted.parquet")

print("CSV successfully converted to Parquet")
```

```python
table = pq.read_table("sample_converted.parquet")
print(table.to_pandas().head())
```

```python
import json
import pyarrow as pa
import pyarrow.parquet as pq

# Load JSON
with open("sample.json", "r", encoding="utf-8") as f:
    data = json.load(f)

# Convert to Arrow table
table = pa.Table.from_pylist(data)

# Write Parquet
pq.write_table(table, "sample_converted1.parquet")

print("JSON successfully converted to Parquet")

table = pq.read_table("sample_converted1.parquet")
print(table.to_pandas().head())
```

```python
# Write to Excel
df.to_excel("sample.xlsx", index=False)

print("Parquet file successfully converted to Excel.")
```
